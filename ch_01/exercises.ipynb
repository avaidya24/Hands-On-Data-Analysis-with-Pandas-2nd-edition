## Exercise 5: Calculating statistics and verifying
### mean

```python
def calc_mean(list_param):
    ls_sum = sum(list_param)
    ls_mean = ls_sum/len(list_param)
    return ls_mean
avg1 = calc_mean(salaries)
print(avg1)
```

### median

```python
def calc_median(list_param):
    # sort the list smallest to largest
    sorted_ls = sorted(list_param)
    # adjust for list lengths with even or odd values
    # even lengths % 2 = 0, odd % 2 = 1
    if len(sorted_ls) % 2 == 0:
        # then it's an even number of items
        m1 = sorted_ls[len(sorted_ls)//2]
        # print(m1)
        m2 = sorted_ls[(len(sorted_ls)//2) - 1]
        # print(m2)
        median = (m1 + m2)/2
        print(median)
    elif len(sorted_ls) % 2 == 1:
        # then it's an odd number of items
        median = sorted_ls[len(sorted_ls)//2]
        print(median)
    else:
        print("ERROR Detected")

calc_median(salaries)
```

### mode

```python
from collections import Counter

def calc_mode(ls_param):
    length = len(ls_param)
    
    data = Counter(ls_param)
    getmode = dict(data)
    mode = [k for k, v in getmode.items() if v == max(list(data.values()))]
    print("Mode: ", mode)
    if len(mode) == length:
        # every value is unique
        getmode1 = "No mode found"
    else:
        getmode1 = "Mode is/are: " + ", ".join(map(str, mode))
    print(getmode1)

calc_mode(salaries)
```

### sample variance
Remember to use Bessel's correction.

```python
def calc_var(ls_param):
    # sort the list
    sort_ls = sorted(ls_param)
    
    # get the mean avg
    ls_mean = calc_mean(ls_param)
    z = sum((i - ls_mean) ** 2 for i in ls_param)/(len(ls_param) - 1)
    
    return z

calc_var(salaries)
```
